## Пример 1.16а

> В аудитории $m=25$ студентов. Найти вероятность того, что хотя бы у двух студентов дни рождения совпадают. При каком числе студентов эта вероятность будет не меньше $0.95$? Считаем все 365 дней года равновозможными.

$P(\overline{A})=\frac{365\cdot364\cdots(365-m+1)}{365^m}$

$P(A)=1-P(\overline{A})$

```python
def birthday_probability(m):
    p_all_different = 1.0

    for i in range(m):
        p_all_different *= (365 - i) / 365

    return 1 - p_all_different

p_25 = birthday_probability(25)
#print(round(p_25, 3))

m = 1
while birthday_probability(m) < 0.95:
    m += 1

print(m)
#print(round(birthday_probability(m), 4))
```

## Задача 4.1. Задача о шляпах

> Гардеробщица одновременно выдала номерки 6 лицам, сдавшим в гардероб свои шляпы, и повесила их наугад. Найти вероятность того, что каждый получит собственную шляпу. Найти вероятность, что каждый получит чужую шляпу.

Всего перестановок: $6!$.

Для второго пункта нужны беспорядки - перестановки, где ни один элемент не стоит на своем месте.

```python
def derangement(n):
    if n == 0:
        return 1
    if n == 1:
        return 0

    d0 = 1
    d1 = 0
    for i in range(2, n + 1):
        d = (i - 1) * (d1 + d0)
        d0 = d1
        d1 = d
    return d1

n = 6
p_own = 1 / math.factorial(n)
p_foreign = derangement(n) / math.factorial(n)

print('Своя шляпа у каждого:', p_own)
print('Чужая шляпа у каждого:', round(p_foreign, 4))
```

## Задача + моделирование 

> Решить классически и смоделировать решение задачи. В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

```python
from math import comb
from random import sample, choice
import matplotlib.pyplot as plt

p = sum(comb(15, k) * comb(5, 3-k) / comb(20, 3) * (6+k)/10 for k in range(4))

urna1 = ['W'] * 15 + ['B'] * 5
urna2 = ['W'] * 6 + ['B']

def experiment():
    balls = sample(urna1, 3)
    return choice(urna2 + balls) == 'W'

N = 100000
success = 0
N_values = []
freq = []

for i in range(1, N + 1):
    if experiment():
        success += 1

    if i % 1000 == 0:
        N_values.append(i)
        freq.append(success / i)

print('Теоретическая вероятность:', round(p, 4))
print('Экспериментальная вероятность:', round(success / N, 4))

plt.plot(N_values, freq, label='Экспериментальная')
plt.axhline(p, color='red', linestyle='--', label='Теоретическая')
plt.xlabel('Количество испытаний')
plt.ylabel('Вероятность')
plt.legend()
plt.grid()
plt.show()
```

## Задача 2.22

> В вузе обучаются 3650 студентов. Вероятность того, что день рождения студента приходится на определенный день года, равна 1/365. Найти:
> - а) наиболее вероятное число студентов, родившихся 1 мая, и вероятность такого события;
> - б) вероятность того, что по крайней мере 3 студента имеют один и тот же день рождения.

Так как $n$ большое, а $p$ маленькое, используем формулу Пуассона.

$\lambda=np=10$.

```python
def poisson_p(k, lam):
    return math.exp(-lam) * lam ** k / math.factorial(k)

n = 3650
p = 1 / 365
lam = n * p

m = 10
p_a = poisson_p(m, lam)
p_b = 1 - sum(poisson_p(k, lam) for k in range(3))

print('а) m =', m, ', P =', round(p_a, 4))
print('б)', round(p_b, 4))
```
